```python
# STEP 12: CONSTRUCT AND SAVE PILOT 4D TENSOR

import os
import numpy as np
import tifffile

# ------------------------------------------
# 1. DEFINE PILOT CROP
# ------------------------------------------

T = 250
Z = 35
H = 96
W = 96

original_height = 512
original_width = 708

y_start = (original_height - H) // 2
x_start = (original_width - W) // 2

y_end = y_start + H
x_end = x_start + W

print("Crop coordinates:")
print("Y:", y_start, "to", y_end)
print("X:", x_start, "to", x_end)

# ------------------------------------------
# 2. CREATE OUTPUT DIRECTORY
# ------------------------------------------

OUTPUT_DIR = "/content/drive/MyDrive/Math_Project/CellDivisionTensor_Results"

os.makedirs(OUTPUT_DIR, exist_ok=True)

tensor_path = os.path.join(
    OUTPUT_DIR,
    "Sequence01_Pilot_4D_Tensor.dat"
)

# ------------------------------------------
# 3. CREATE MEMORY-MAPPED TENSOR
# ------------------------------------------

tensor_4d = np.memmap(
    tensor_path,
    dtype=np.uint8,
    mode="w+",
    shape=(T, Z, H, W)
)

# ------------------------------------------
# 4. LOAD AND CROP ALL 250 FRAMES
# ------------------------------------------

for t in range(T):

    filename = f"t{t:03d}.tif"

    volume = tifffile.imread(
        os.path.join(DATA_PATH, filename)
    )

    if volume.shape != (35, 512, 708):
        raise ValueError(
            f"Unexpected shape at frame {t}: {volume.shape}"
        )

    tensor_4d[t] = volume[
        :,
        y_start:y_end,
        x_start:x_end
    ]

    if (t + 1) % 25 == 0:
        print(f"Processed {t + 1}/{T} frames")

# ------------------------------------------
# 5. FLUSH AND VERIFY
# ------------------------------------------

tensor_4d.flush()

print("\nPILOT TENSOR CREATED SUCCESSFULLY")
print("Shape:", tensor_4d.shape)
print("Data type:", tensor_4d.dtype)
print("File size (MB):", round(os.path.getsize(tensor_path) / (1024**2), 2))
print("Minimum:", tensor_4d.min())
print("Maximum:", tensor_4d.max())
print("Saved at:", tensor_path)
```

```python
# STEP 15: MODE 1 (TIME) SVD

import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.linalg import svd

# Convert the tensor to float32 for decomposition
tensor_float = np.asarray(tensor_4d, dtype=np.float32)

# Mode 1 unfolding: Time
mode1_matrix = tensor_float.reshape(250, -1)

print("Mode 1 unfolding shape:", mode1_matrix.shape)

# Calculate SVD
U1_full, S1, V1t = svd(
    mode1_matrix,
    full_matrices=False,
    check_finite=False,
    lapack_driver="gesdd"
)

# Save results
np.save(os.path.join(OUTPUT_DIR, "U1_full.npy"), U1_full)
np.save(os.path.join(OUTPUT_DIR, "singular_values_mode_1.npy"), S1)

# Plot singular values
plt.figure(figsize=(10, 5))
plt.plot(
    np.arange(1, len(S1) + 1),
    S1,
    marker="o",
    markersize=3
)

plt.xlabel("Singular Value Index")
plt.ylabel("Singular Value")
plt.title("Mode 1 - Temporal Singular Value Distribution")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "Step15_Mode1_Singular_Values.png"),
    dpi=300
)

plt.show()

# Energy retention
energy = np.cumsum(S1**2) / np.sum(S1**2)

plt.figure(figsize=(10, 5))
plt.plot(
    np.arange(1, len(S1) + 1),
    energy * 100
)

plt.axhline(90, linestyle="--", label="90%")
plt.axhline(95, linestyle="--", label="95%")
plt.axhline(99, linestyle="--", label="99%")

plt.xlabel("Rank")
plt.ylabel("Cumulative Energy (%)")
plt.title("Mode 1 - Temporal Energy Retention")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "Step15_Mode1_Energy.png"),
    dpi=300
)

plt.show()

print("\nTEMPORAL RANK REQUIREMENTS")

for threshold in [0.90, 0.95, 0.99]:
    rank = np.searchsorted(energy, threshold) + 1
    print(f"{threshold*100:.0f}% energy: Rank {rank}")
```

```python

# STEP 16: MODE 2 (DEPTH) SVD

import numpy as np
import matplotlib.pyplot as plt
import os

# Dimensions
T, Z, Y, X = 250, 35, 96, 96

print("Mode 2 unfolding shape:", (Z, T * Y * X))

# Calculate the Gram matrix without creating the
# complete unfolding matrix in memory.
gram_matrix = np.zeros((Z, Z), dtype=np.float64)

for t in range(T):
    frame = np.asarray(tensor_4d[t], dtype=np.float32)
    depth_matrix = frame.reshape(Z, Y * X)

    gram_matrix += depth_matrix @ depth_matrix.T

# Eigenvalue decomposition of the Gram matrix
eigenvalues, eigenvectors = np.linalg.eigh(gram_matrix)

# Sort eigenvalues and eigenvectors in descending order
order = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

# Singular values and left singular vectors
S2 = np.sqrt(np.maximum(eigenvalues, 0))
U2_full = eigenvectors.astype(np.float32)

# Save results
np.save(os.path.join(OUTPUT_DIR, "U2_full.npy"), U2_full)
np.save(os.path.join(OUTPUT_DIR, "singular_values_mode_2.npy"), S2)

# Calculate cumulative energy
energy = S2**2
cumulative_energy = np.cumsum(energy) / np.sum(energy) * 100

# Calculate required ranks
thresholds = [90, 95, 99]

print("\nDEPTH RANK REQUIREMENTS")

for threshold in thresholds:
    rank = np.searchsorted(cumulative_energy, threshold) + 1
    print(f"{threshold}% energy: Rank {rank}")

# Plot singular values
plt.figure(figsize=(10, 5))
plt.plot(np.arange(1, len(S2)+1), S2, marker=".")
plt.title("Mode 2 - Depth Singular Value Distribution")
plt.xlabel("Singular Value Index")
plt.ylabel("Singular Value")
plt.grid(True)
plt.savefig(os.path.join(OUTPUT_DIR, "Mode2_Singular_Values.png"),
            dpi=300, bbox_inches="tight")
plt.show()

# Plot cumulative energy
plt.figure(figsize=(10, 5))
plt.plot(np.arange(1, len(S2)+1), cumulative_energy)
for threshold in thresholds:
    plt.axhline(threshold, linestyle="--", label=f"{threshold}%")

plt.title("Mode 2 - Depth Energy Retention")
plt.xlabel("Rank")
plt.ylabel("Cumulative Energy (%)")
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(OUTPUT_DIR, "Mode2_Cumulative_Energy.png"),
            dpi=300, bbox_inches="tight")
plt.show()

print("\nMode 2 completed.")
print("U2 shape:", U2_full.shape)
print("Singular values:", S2.shape)
```

```python

# STEP 21: QUANTITATIVE HOSVD RECONSTRUCTION EVALUATION

import numpy as np
import matplotlib.pyplot as plt
import os
import gc

# Open original and reconstructed tensors
original = tensor_4d

reconstruction = np.memmap(
    os.path.join(OUTPUT_DIR, "HOSVD_Reconstruction_95.dat"),
    dtype=np.float32,
    mode="r",
    shape=(250, 35, 96, 96)
)

# Accumulators for reconstruction metrics
squared_error = 0.0
absolute_error = 0.0
original_squared_sum = 0.0

original_min = np.inf
original_max = -np.inf

original_frame_means = []
reconstructed_frame_means = []

original_temporal_changes = []
reconstructed_temporal_changes = []

# Process one frame at a time
for t in range(250):

    x = np.asarray(original[t], dtype=np.float32)
    x_hat = np.asarray(reconstruction[t], dtype=np.float32)

    difference = x - x_hat

    squared_error += np.sum(difference ** 2, dtype=np.float64)
    absolute_error += np.sum(np.abs(difference), dtype=np.float64)
    original_squared_sum += np.sum(x ** 2, dtype=np.float64)

    original_min = min(original_min, float(x.min()))
    original_max = max(original_max, float(x.max()))

    original_frame_means.append(float(x.mean()))
    reconstructed_frame_means.append(float(x_hat.mean()))

    # Frame-to-frame temporal changes
    if t > 0:
        previous_x = np.asarray(original[t-1], dtype=np.float32)
        previous_hat = np.asarray(reconstruction[t-1], dtype=np.float32)

        original_temporal_changes.append(
            float(np.mean(np.abs(x - previous_x)))
        )

        reconstructed_temporal_changes.append(
            float(np.mean(np.abs(x_hat - previous_hat)))
        )

    if (t + 1) % 50 == 0:
        print(f"Evaluated {t+1}/250 frames")

# Calculate reconstruction metrics
total_elements = 250 * 35 * 96 * 96

mse = squared_error / total_elements
rmse = np.sqrt(mse)
mae = absolute_error / total_elements

relative_error = (
    np.sqrt(squared_error) /
    np.sqrt(original_squared_sum)
)

data_range = original_max - original_min

psnr = (
    20 * np.log10(data_range / rmse)
    if rmse > 0 and data_range > 0
    else np.inf
)

# Temporal preservation metrics
original_temporal_changes = np.array(original_temporal_changes)
reconstructed_temporal_changes = np.array(reconstructed_temporal_changes)

temporal_mae = np.mean(
    np.abs(original_temporal_changes - reconstructed_temporal_changes)
)

original_mean_change = np.mean(original_temporal_changes)
reconstructed_mean_change = np.mean(reconstructed_temporal_changes)

temporal_relative_error = (
    temporal_mae / original_mean_change
    if original_mean_change > 0
    else np.nan
)

# Print results
print("\n" + "=" * 55)
print("HOSVD RECONSTRUCTION EVALUATION")
print("=" * 55)

print(f"Original minimum: {original_min:.4f}")
print(f"Original maximum: {original_max:.4f}")
print(f"MSE: {mse:.6f}")
print(f"RMSE: {rmse:.6f}")
print(f"MAE: {mae:.6f}")
print(f"Relative Frobenius error: {relative_error:.6f}")
print(f"Relative error (%): {relative_error * 100:.4f}%")
print(f"PSNR: {psnr:.4f} dB")

print("\nTEMPORAL PRESERVATION")
print(f"Original mean frame change: {original_mean_change:.6f}")
print(f"Reconstructed mean frame change: {reconstructed_mean_change:.6f}")
print(f"Temporal change MAE: {temporal_mae:.6f}")
print(f"Temporal relative error: {temporal_relative_error:.6f}")

# Save numerical results
metrics_path = os.path.join(
    OUTPUT_DIR, "HOSVD_Reconstruction_Metrics_95.txt"
)

with open(metrics_path, "w") as f:
    f.write("HOSVD RECONSTRUCTION METRICS\n")
    f.write(f"MSE: {mse}\n")
    f.write(f"RMSE: {rmse}\n")
    f.write(f"MAE: {mae}\n")
    f.write(f"Relative Frobenius error: {relative_error}\n")
    f.write(f"PSNR: {psnr}\n")
    f.write(f"Temporal change MAE: {temporal_mae}\n")
    f.write(f"Temporal relative error: {temporal_relative_error}\n")

# GRAPH 1: Mean intensity comparison
plt.figure(figsize=(12, 5))
plt.plot(original_frame_means, label="Original")
plt.plot(reconstructed_frame_means, label="HOSVD reconstructed")
plt.title("Original vs Reconstructed Mean Intensity")
plt.xlabel("Frame")
plt.ylabel("Mean intensity")
plt.legend()
plt.grid(True)
plt.savefig(
    os.path.join(OUTPUT_DIR, "HOSVD_Mean_Intensity_Comparison.png"),
    dpi=300, bbox_inches="tight"
)
plt.show()

# GRAPH 2: Temporal change comparison
plt.figure(figsize=(12, 5))
plt.plot(original_temporal_changes, label="Original")
plt.plot(reconstructed_temporal_changes, label="HOSVD reconstructed")
plt.title("Original vs Reconstructed Temporal Changes")
plt.xlabel("Frame transition")
plt.ylabel("Mean absolute voxel change")
plt.legend()
plt.grid(True)
plt.savefig(
    os.path.join(OUTPUT_DIR, "HOSVD_Temporal_Change_Comparison.png"),
    dpi=300, bbox_inches="tight"
)
plt.show()

# GRAPH 3: Example frame comparison
sample_frame = 100
sample_z = 17

x = np.asarray(original[sample_frame, sample_z], dtype=np.float32)
x_hat = np.asarray(reconstruction[sample_frame, sample_z], dtype=np.float32)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(x, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(x_hat, cmap="gray")
plt.title("HOSVD Reconstruction")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(np.abs(x - x_hat), cmap="hot")
plt.title("Absolute Reconstruction Error")
plt.colorbar()
plt.axis("off")

plt.suptitle(f"Frame {sample_frame}, Depth {sample_z}")
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "HOSVD_Frame_Comparison.png"),
    dpi=300, bbox_inches="tight"
)
plt.show()

print("\nAll metrics and graphs saved to:", OUTPUT_DIR)

del x, x_hat
gc.collect()
```

```python

# STEP 25: CORRECT RECONSTRUCTION FOR BOTH HOSVD EXPERIMENTS

import numpy as np
import os
import gc

def reconstruct_hosvd(core, U1, U2, U3, U4, output_name):

    print("\nStarting:", output_name)

    # Expand Mode 4: Width
    result = np.tensordot(
        core, U4.T, axes=(3, 0)
    )
    print("After Mode 4:", result.shape)

    # Expand Mode 3: Height
    result = np.tensordot(
        result, U3.T, axes=(2, 0)
    )

    # IMPORTANT: Restore height to axis 2
    result = np.moveaxis(result, -1, 2)

    print("After Mode 3:", result.shape)

    # Expand Mode 2: Depth
    result = np.tensordot(
        result, U2.T, axes=(1, 0)
    )

    # Restore depth to axis 1
    result = np.moveaxis(result, -1, 1)

    print("After Mode 2:", result.shape)

    # Validate before expanding time
    expected = (U1.shape[0], U2.shape[1],
                U3.shape[1], U4.shape[1])

    assert result.shape == (
        U1.shape[1], U2.shape[0],
        U3.shape[0], U4.shape[0]
    )

    # Create output file
    path = os.path.join(OUTPUT_DIR, output_name)

    output = np.memmap(
        path,
        dtype=np.float32,
        mode="w+",
        shape=(250, 35, 96, 96)
    )

    # Expand Mode 1: Time
    for t in range(250):

        output[t] = np.tensordot(
            U1[t], result, axes=(0, 0)
        )

        if (t + 1) % 50 == 0:
            print(f"Reconstructed {t+1}/250 frames")

    output.flush()

    print("Completed:", output_name)
    print("Final shape:", output.shape)

    del result, output
    gc.collect()

# Load 95% factors
core_95 = np.load(
    os.path.join(OUTPUT_DIR, "HOSVD_Core_95.npy")
)

factors_95 = np.load(
    os.path.join(OUTPUT_DIR, "HOSVD_Factors_95.npz")
)

# Reconstruct 95%
reconstruct_hosvd(
    core_95,
    factors_95["U1"],
    factors_95["U2"],
    factors_95["U3"],
    factors_95["U4"],
    "HOSVD_Reconstruction_95_Corrected.dat"
)

del core_95, factors_95
gc.collect()

# Load 99% factors
core_99 = np.load(
    os.path.join(OUTPUT_DIR, "HOSVD_Core_99.npy")
)

factors_99 = np.load(
    os.path.join(OUTPUT_DIR, "HOSVD_Factors_99.npz")
)

# Reconstruct 99%
reconstruct_hosvd(
    core_99,
    factors_99["U1"],
    factors_99["U2"],
    factors_99["U3"],
    factors_99["U4"],
    "HOSVD_Reconstruction_99_Corrected.dat"
)

del core_99, factors_99
gc.collect()

print("\nBOTH CORRECTED RECONSTRUCTIONS COMPLETED.")
```

```python

# STEP 26: EVALUATE CORRECTED HOSVD RECONSTRUCTIONS

import numpy as np
import os

# File paths
original_path = os.path.join(OUTPUT_DIR, "Sequence01_Pilot_4D_Tensor.dat")
path_95 = os.path.join(OUTPUT_DIR, "HOSVD_Reconstruction_95_Corrected.dat")
path_99 = os.path.join(OUTPUT_DIR, "HOSVD_Reconstruction_99_Corrected.dat")

shape = (250, 35, 96, 96)
total_elements = np.prod(shape)
data_range = 243.0

# Memory-mapped arrays
original = np.memmap(original_path, dtype=np.uint8, mode="r", shape=shape)
recon_95 = np.memmap(path_95, dtype=np.float32, mode="r", shape=shape)
recon_99 = np.memmap(path_99, dtype=np.float32, mode="r", shape=shape)

# Accumulators
results = {}
temporal_changes = {
    "Original": [],
    "95% HOSVD": [],
    "99% HOSVD": []
}

for name, reconstruction in [
    ("95% HOSVD", recon_95),
    ("99% HOSVD", recon_99)
]:
    squared_error_sum = 0.0
    absolute_error_sum = 0.0
    original_squared_sum = 0.0

    for t in range(shape[0]):
        x = original[t].astype(np.float64)
        y = reconstruction[t].astype(np.float64)

        error = x - y

        squared_error_sum += np.sum(error ** 2)
        absolute_error_sum += np.sum(np.abs(error))
        original_squared_sum += np.sum(x ** 2)

        if t > 0:
            original_change = np.mean(
                np.abs(original[t].astype(np.float32) -
                       original[t-1].astype(np.float32))
            )

            reconstructed_change = np.mean(
                np.abs(reconstruction[t] - reconstruction[t-1])
            )

            if name == "95% HOSVD":
                temporal_changes["Original"].append(original_change)

            temporal_changes[name].append(reconstructed_change)

    mse = squared_error_sum / total_elements
    rmse = np.sqrt(mse)
    mae = absolute_error_sum / total_elements
    relative_error = (
        np.sqrt(squared_error_sum) /
        np.sqrt(original_squared_sum)
    ) * 100

    psnr = 20 * np.log10(data_range / rmse)

    results[name] = {
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae,
        "Relative Frobenius Error (%)": relative_error,
        "PSNR (dB)": psnr
    }

# Temporal preservation metrics
original_temporal = np.array(temporal_changes["Original"])

for name in ["95% HOSVD", "99% HOSVD"]:
    reconstructed_temporal = np.array(temporal_changes[name])

    temporal_mae = np.mean(
        np.abs(original_temporal - reconstructed_temporal)
    )

    temporal_relative_error = (
        temporal_mae / np.mean(original_temporal)
    ) * 100

    results[name]["Mean Temporal Change"] = np.mean(
        reconstructed_temporal
    )
    results[name]["Temporal MAE"] = temporal_mae
    results[name]["Temporal Relative Error (%)"] = temporal_relative_error

# Display results
print("=" * 65)
print("CORRECTED HOSVD RECONSTRUCTION RESULTS")
print("=" * 65)

for name, metrics in results.items():
    print(f"\n{name}")
    print("-" * 45)
    for metric, value in metrics.items():
        print(f"{metric}: {value:.6f}")

print("\nOriginal mean temporal change:",
      np.mean(original_temporal))

# Save results
np.savez(
    os.path.join(OUTPUT_DIR, "Corrected_HOSVD_Metrics.npz"),
    results_95=results["95% HOSVD"],
    results_99=results["99% HOSVD"],
    original_temporal=original_temporal,
    temporal_95=np.array(temporal_changes["95% HOSVD"]),
    temporal_99=np.array(temporal_changes["99% HOSVD"])
)

print("\nMetrics saved successfully.")
```

```python

# STEP 28: EXTRACT CANDIDATE CELL-DIVISION EVENTS FROM TRACK METADATA

import pandas as pd
import numpy as np
import os

track_file = os.path.join(BASE, "01_GT", "TRA", "man_track.txt")

# Cell Tracking Challenge track metadata has four columns and no header
tracks = pd.read_csv(
    track_file,
    sep=r"\s+",
    header=None,
    names=["Track_ID", "Start", "End", "Parent_ID"]
)

# Ensure numeric values
tracks = tracks.astype({
    "Track_ID": int,
    "Start": int,
    "End": int,
    "Parent_ID": int
})

# Group child tracks by their parent
children_by_parent = (
    tracks[tracks["Parent_ID"] != 0]
    .groupby("Parent_ID")["Track_ID"]
    .apply(list)
)

# Keep parents with exactly two recorded children
division_parents = children_by_parent[
    children_by_parent.apply(len) == 2
]

track_lookup = tracks.set_index("Track_ID")

division_events = []

for parent_id, child_ids in division_parents.items():

    if parent_id not in track_lookup.index:
        continue

    parent = track_lookup.loc[parent_id]
    child_1 = track_lookup.loc[child_ids[0]]
    child_2 = track_lookup.loc[child_ids[1]]

    # Candidate event frame: earliest recorded child start
    event_frame = min(child_1["Start"], child_2["Start"])

    division_events.append({
        "Parent_ID": parent_id,
        "Parent_Start": parent["Start"],
        "Parent_End": parent["End"],
        "Child_1_ID": child_ids[0],
        "Child_1_Start": child_1["Start"],
        "Child_2_ID": child_ids[1],
        "Child_2_Start": child_2["Start"],
        "Candidate_Event_Frame": event_frame,
        "Children_Start_Match": (
            child_1["Start"] == child_2["Start"]
        )
    })

division_df = pd.DataFrame(division_events)

# The pilot has 250 frames, but tracking annotations end at frame 194
division_df = division_df[
    division_df["Candidate_Event_Frame"].between(0, 194)
].copy()

division_df = division_df.sort_values(
    "Candidate_Event_Frame"
).reset_index(drop=True)

print("=" * 65)
print("CANDIDATE CELL-DIVISION EVENT SUMMARY")
print("=" * 65)

print("Total tracks:", len(tracks))
print("Parents with exactly two children:", len(division_parents))
print("Candidate events within annotated frames:", len(division_df))
print(
    "Events where both children start in the same frame:",
    division_df["Children_Start_Match"].sum()
)

print("\nFirst 15 candidate events:")
display(division_df.head(15))

# Save the event table
event_path = os.path.join(
    OUTPUT_DIR, "STEP28_Candidate_Division_Events.csv"
)

division_df.to_csv(event_path, index=False)

print("\nEvent table saved:", event_path)
```

```python

# STEP 29: VERIFY PARENT AND CHILD TRACK LABELS IN THE PILOT CROP

import os
import numpy as np
import pandas as pd
import tifffile

TRA_DIR = os.path.join(BASE, "01_GT", "TRA")

# Same spatial crop used to create the pilot tensor
Y_START, Y_END = 208, 304
X_START, X_END = 306, 402

# Build a list of label checks required for each frame
frame_checks = {}

for idx, event in division_df.iterrows():
    event_frame = int(event["Candidate_Event_Frame"])

    checks = [
        (event_frame - 1, idx, int(event["Parent_ID"]), "parent"),
        (event_frame, idx, int(event["Child_1_ID"]), "child1"),
        (event_frame + 1, idx, int(event["Child_1_ID"]), "child1"),
        (event_frame, idx, int(event["Child_2_ID"]), "child2"),
        (event_frame + 1, idx, int(event["Child_2_ID"]), "child2")
    ]

    for frame, event_idx, track_id, role in checks:
        if 0 <= frame <= 194:
            frame_checks.setdefault(frame, []).append(
                (event_idx, track_id, role)
            )

# Store pixel counts for every requested track/frame combination
pixel_counts = {}

for frame in sorted(frame_checks):
    mask_path = os.path.join(
        TRA_DIR, f"man_track{frame:03d}.tif"
    )

    mask = tifffile.imread(mask_path)

    if mask.shape != (35, 512, 708):
        raise ValueError(
            f"Unexpected mask shape at frame {frame}: {mask.shape}"
        )

    crop = mask[
        :,
        Y_START:Y_END,
        X_START:X_END
    ]

    for event_idx, track_id, role in frame_checks[frame]:
        count = np.count_nonzero(crop == track_id)

        pixel_counts[(event_idx, frame, role)] = count

# Summarize the checks for each candidate event
verification_rows = []

for idx, event in division_df.iterrows():
    event_frame = int(event["Candidate_Event_Frame"])

    parent_frame = event_frame - 1

    parent_count = pixel_counts.get(
        (idx, parent_frame, "parent"), 0
    )

    child1_count = sum(
        pixel_counts.get((idx, f, "child1"), 0)
        for f in [event_frame, event_frame + 1]
    )

    child2_count = sum(
        pixel_counts.get((idx, f, "child2"), 0)
        for f in [event_frame, event_frame + 1]
    )

    verification_rows.append({
        "Parent_ID": int(event["Parent_ID"]),
        "Candidate_Event_Frame": event_frame,
        "Parent_Pixels_Before": parent_count,
        "Child1_Pixels_After": child1_count,
        "Child2_Pixels_After": child2_count,
        "Parent_Visible": parent_count > 0,
        "Child1_Visible": child1_count > 0,
        "Child2_Visible": child2_count > 0,
        "All_Three_Visible": (
            parent_count > 0
            and child1_count > 0
            and child2_count > 0
        )
    })

verification_df = pd.DataFrame(verification_rows)

# Summarize crop coverage
print("=" * 65)
print("PILOT CROP DIVISION-ANNOTATION VERIFICATION")
print("=" * 65)

print("Total candidate events:", len(verification_df))
print(
    "Parents visible before event:",
    verification_df["Parent_Visible"].sum()
)
print(
    "Child 1 visible after event:",
    verification_df["Child1_Visible"].sum()
)
print(
    "Child 2 visible after event:",
    verification_df["Child2_Visible"].sum()
)
print(
    "Events with all three tracks visible:",
    verification_df["All_Three_Visible"].sum()
)

print("\nFirst 20 verification results:")
display(verification_df.head(20))

# Save verification table
verification_path = os.path.join(
    OUTPUT_DIR, "STEP29_Division_Crop_Verification.csv"
)

verification_df.to_csv(verification_path, index=False)

print("\nVerification table saved:", verification_path)
```

```python

# STEP 30: EXTRACT LOCAL REGIONS FOR VERIFIED DIVISION EVENTS

import os
import numpy as np
import pandas as pd
import tifffile

TRA_DIR = os.path.join(BASE, "01_GT", "TRA")

# Select only events where parent and both children were detected
verified_events = verification_df[
    verification_df["All_Three_Visible"] == True
].copy().reset_index(drop=True)

# Reload the corresponding event metadata
verified_events = verified_events.merge(
    division_df[
        [
            "Parent_ID",
            "Parent_End",
            "Child_1_ID",
            "Child_2_ID",
            "Candidate_Event_Frame"
        ]
    ],
    on=["Parent_ID", "Candidate_Event_Frame"],
    how="left"
)

Y_START, Y_END = 208, 304
X_START, X_END = 306, 402

event_regions = []

for _, event in verified_events.iterrows():

    parent_id = int(event["Parent_ID"])
    child1_id = int(event["Child_1_ID"])
    child2_id = int(event["Child_2_ID"])
    frame = int(event["Candidate_Event_Frame"])

    # Collect coordinates from parent-before and both child-after masks
    coordinate_sets = []

    checks = [
        (frame - 1, parent_id),
        (frame, child1_id),
        (frame + 1, child1_id),
        (frame, child2_id),
        (frame + 1, child2_id)
    ]

    for mask_frame, track_id in checks:

        mask_path = os.path.join(
            TRA_DIR, f"man_track{mask_frame:03d}.tif"
        )

        mask = tifffile.imread(mask_path)

        crop = mask[
            :,
            Y_START:Y_END,
            X_START:X_END
        ]

        coords = np.argwhere(crop == track_id)

        if len(coords) > 0:
            coordinate_sets.append(coords)

    if not coordinate_sets:
        continue

    all_coords = np.concatenate(coordinate_sets, axis=0)

    # Bounding box in pilot-crop coordinates
    z_min, y_min, x_min = all_coords.min(axis=0)
    z_max, y_max, x_max = all_coords.max(axis=0)

    # Add a small spatial margin, staying inside the pilot crop
    margin = 5

    y0 = max(0, int(y_min) - margin)
    y1 = min(96, int(y_max) + margin + 1)
    x0 = max(0, int(x_min) - margin)
    x1 = min(96, int(x_max) + margin + 1)

    event_regions.append({
        "Parent_ID": parent_id,
        "Event_Frame": frame,
        "Child_1_ID": child1_id,
        "Child_2_ID": child2_id,
        "Z_Min": int(z_min),
        "Z_Max": int(z_max),
        "Y0": y0,
        "Y1": y1,
        "X0": x0,
        "X1": x1,
        "ROI_Height": y1 - y0,
        "ROI_Width": x1 - x0
    })

event_regions_df = pd.DataFrame(event_regions)

print("=" * 65)
print("VERIFIED DIVISION EVENT REGIONS")
print("=" * 65)

print("Events with extracted regions:", len(event_regions_df))

if len(event_regions_df) > 0:
    print("\nRegion dimensions:")
    print(
        event_regions_df[
            ["ROI_Height", "ROI_Width"]
        ].describe()
    )

    print("\nExtracted event regions:")
    display(event_regions_df)

# Save event region table
region_path = os.path.join(
    OUTPUT_DIR, "STEP30_Verified_Division_Regions.csv"
)

event_regions_df.to_csv(region_path, index=False)

print("\nRegion table saved:", region_path)
```

```python

# STEP 31: EVALUATE HOSVD RECONSTRUCTION INSIDE DIVISION REGIONS

import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt

# Load corrected reconstruction files
recon_95 = np.memmap(
    os.path.join(OUTPUT_DIR, "HOSVD_Reconstruction_95_Corrected.dat"),
    dtype=np.float32,
    mode="r",
    shape=(250, 35, 96, 96)
)

recon_99 = np.memmap(
    os.path.join(OUTPUT_DIR, "HOSVD_Reconstruction_99_Corrected.dat"),
    dtype=np.float32,
    mode="r",
    shape=(250, 35, 96, 96)
)

event_metrics = []

# Accumulate squared and absolute errors for overall weighted metrics
total_squared_95 = 0.0
total_squared_99 = 0.0
total_absolute_95 = 0.0
total_absolute_99 = 0.0
total_voxels = 0

for _, event in event_regions_df.iterrows():

    event_frame = int(event["Event_Frame"])

    # Three-frame temporal window around the candidate division
    t0 = event_frame - 1
    t1 = event_frame + 2

    z0 = int(event["Z_Min"])
    z1 = int(event["Z_Max"]) + 1

    y0, y1 = int(event["Y0"]), int(event["Y1"])
    x0, x1 = int(event["X0"]), int(event["X1"])

    original_roi = tensor_4d[
        t0:t1, z0:z1, y0:y1, x0:x1
    ].astype(np.float32)

    roi_95 = recon_95[
        t0:t1, z0:z1, y0:y1, x0:x1
    ]

    roi_99 = recon_99[
        t0:t1, z0:z1, y0:y1, x0:x1
    ]

    error_95 = original_roi - roi_95
    error_99 = original_roi - roi_99

    rmse_95 = np.sqrt(np.mean(error_95 ** 2))
    rmse_99 = np.sqrt(np.mean(error_99 ** 2))

    mae_95 = np.mean(np.abs(error_95))
    mae_99 = np.mean(np.abs(error_99))

    event_metrics.append({
        "Parent_ID": int(event["Parent_ID"]),
        "Event_Frame": event_frame,
        "ROI_Voxels": original_roi.size,
        "RMSE_95": rmse_95,
        "RMSE_99": rmse_99,
        "MAE_95": mae_95,
        "MAE_99": mae_99
    })

    total_squared_95 += np.sum(error_95 ** 2)
    total_squared_99 += np.sum(error_99 ** 2)

    total_absolute_95 += np.sum(np.abs(error_95))
    total_absolute_99 += np.sum(np.abs(error_99))

    total_voxels += original_roi.size

event_metrics_df = pd.DataFrame(event_metrics)

# Overall voxel-weighted metrics across the 14 event regions
overall_rmse_95 = np.sqrt(total_squared_95 / total_voxels)
overall_rmse_99 = np.sqrt(total_squared_99 / total_voxels)

overall_mae_95 = total_absolute_95 / total_voxels
overall_mae_99 = total_absolute_99 / total_voxels

print("=" * 65)
print("DIVISION-REGION RECONSTRUCTION RESULTS")
print("=" * 65)

print("Number of evaluated events:", len(event_metrics_df))
print("Total evaluated voxels:", total_voxels)

print("\n95% HOSVD")
print("Weighted RMSE:", overall_rmse_95)
print("Weighted MAE:", overall_mae_95)

print("\n99% HOSVD")
print("Weighted RMSE:", overall_rmse_99)
print("Weighted MAE:", overall_mae_99)

print("\nAverage per-event RMSE:")
print("95%:", event_metrics_df["RMSE_95"].mean())
print("99%:", event_metrics_df["RMSE_99"].mean())

print("\nAverage per-event MAE:")
print("95%:", event_metrics_df["MAE_95"].mean())
print("99%:", event_metrics_df["MAE_99"].mean())

print("\nPer-event results:")
display(event_metrics_df)

# Save results
metrics_path = os.path.join(
    OUTPUT_DIR, "STEP31_Division_Region_Metrics.csv"
)

event_metrics_df.to_csv(metrics_path, index=False)

# Plot per-event RMSE comparison
positions = np.arange(len(event_metrics_df))
width = 0.38

plt.figure(figsize=(13, 5))

plt.bar(
    positions - width / 2,
    event_metrics_df["RMSE_95"],
    width,
    label="HOSVD 95%"
)

plt.bar(
    positions + width / 2,
    event_metrics_df["RMSE_99"],
    width,
    label="HOSVD 99%"
)

plt.xlabel("Candidate Division Event")
plt.ylabel("Local RMSE")
plt.title("Reconstruction Error Around Annotated Division Events")
plt.xticks(
    positions,
    event_metrics_df["Parent_ID"],
    rotation=45
)
plt.legend()
plt.tight_layout()

plot_path = os.path.join(
    OUTPUT_DIR, "STEP31_Division_Region_RMSE.png"
)

plt.savefig(plot_path, dpi=300, bbox_inches="tight")
plt.show()

print("\nMetrics saved:", metrics_path)
print("Plot saved:", plot_path)
```

```python

# STEP 34: EXTRACT TRACK-LEVEL FEATURES FOR SVM

import os
import numpy as np
import pandas as pd
import tifffile

# Verify required variables
required_vars = ["tensor_4d", "visible_df", "svm_audit_df", "OUTPUT_DIR", "BASE"]
missing = [v for v in required_vars if v not in globals()]

if missing:
    raise NameError(f"Missing required variables: {missing}")

# Pilot crop coordinates used to create tensor_4d
Y0, Y1 = 208, 304
X0, X1 = 306, 402

# Load the track metadata
track_metadata = pd.read_csv(
    os.path.join(BASE, "01_GT", "TRA", "man_track.txt"),
    sep=r"\s+",
    header=None,
    names=["Track_ID", "Start", "End", "Parent_ID"]
)

# Use only tracks visible in the pilot crop
visible_tracks = visible_df["Track_ID"].astype(int).tolist()

feature_rows = []

for track_id in visible_tracks:

    metadata = track_metadata[
        track_metadata["Track_ID"] == track_id
    ].iloc[0]

    start = int(metadata["Start"])
    end = int(metadata["End"])

    frame_mean_intensities = []
    frame_voxel_counts = []
    track_intensities = []

    # Read the annotated mask for each frame in this track
    for frame in range(start, end + 1):

        mask_path = os.path.join(
            BASE, "01_GT", "TRA",
            f"man_track{frame:03d}.tif"
        )

        if not os.path.exists(mask_path):
            continue

        mask = tifffile.imread(mask_path)
        cropped_mask = mask[:, Y0:Y1, X0:X1]

        track_mask = cropped_mask == track_id
        voxel_count = int(np.count_nonzero(track_mask))

        if voxel_count == 0:
            continue

        image = np.asarray(tensor_4d[frame], dtype=np.float32)

        intensities = image[track_mask]

        frame_voxel_counts.append(voxel_count)
        frame_mean_intensities.append(float(np.mean(intensities)))
        track_intensities.extend(intensities.tolist())

    # Skip tracks with no usable observations
    if len(track_intensities) == 0:
        continue

    frame_means = np.asarray(frame_mean_intensities, dtype=np.float32)
    voxel_counts = np.asarray(frame_voxel_counts, dtype=np.float32)
    all_intensities = np.asarray(track_intensities, dtype=np.float32)

    # Consecutive-frame intensity variation
    if len(frame_means) > 1:
        temporal_changes = np.abs(np.diff(frame_means))
        mean_temporal_change = float(np.mean(temporal_changes))
    else:
        mean_temporal_change = 0.0

    # Obtain the audited binary label
    label = int(
        svm_audit_df.loc[
            svm_audit_df["Track_ID"] == track_id,
            "Label"
        ].iloc[0]
    )

    feature_rows.append({
        "Track_ID": track_id,
        "Label": label,
        "Duration": int(metadata["End"] - metadata["Start"] + 1),
        "Observed_Frames": len(frame_means),
        "Mean_Intensity": float(np.mean(all_intensities)),
        "Std_Intensity": float(np.std(all_intensities)),
        "Min_Intensity": float(np.min(all_intensities)),
        "Max_Intensity": float(np.max(all_intensities)),
        "Mean_Frame_Intensity": float(np.mean(frame_means)),
        "Std_Frame_Intensity": float(np.std(frame_means)),
        "Mean_Temporal_Change": mean_temporal_change,
        "Mean_Voxels_Per_Frame": float(np.mean(voxel_counts)),
        "Std_Voxels_Per_Frame": float(np.std(voxel_counts)),
        "Total_Observed_Voxels": int(np.sum(voxel_counts))
    })

# Create feature dataset
svm_features_df = pd.DataFrame(feature_rows)

# Save results
feature_path = os.path.join(
    OUTPUT_DIR,
    "STEP34_SVM_Track_Features.csv"
)

svm_features_df.to_csv(feature_path, index=False)

print("=" * 65)
print("STEP 34: SVM FEATURE EXTRACTION SUMMARY")
print("=" * 65)

print("Visible tracks audited:", len(visible_tracks))
print("Tracks with extracted features:", len(svm_features_df))
print("Feature columns:", len(svm_features_df.columns) - 2)

print("\nClass distribution:")
print(svm_features_df["Label"].value_counts().sort_index())

print("\nFeature dataset preview:")
display(svm_features_df.head(10))

print("\nMissing values:")
print(svm_features_df.isnull().sum())

print("\nSaved:", feature_path)
```

```python

# STEP 35: EXTRACT SVM FEATURES FROM HOSVD RECONSTRUCTIONS

import os
import numpy as np
import pandas as pd
import tifffile

# --------------------------------------------------
# 1. VERIFY INPUTS AND LOAD RECONSTRUCTIONS
# --------------------------------------------------

required_vars = ["visible_df", "svm_audit_df", "OUTPUT_DIR", "BASE"]

missing = [v for v in required_vars if v not in globals()]
if missing:
    raise NameError(f"Missing variables: {missing}")

shape = (250, 35, 96, 96)
dtype = np.float32

path_95 = os.path.join(
    OUTPUT_DIR, "HOSVD_Reconstruction_95_Corrected.dat"
)
path_99 = os.path.join(
    OUTPUT_DIR, "HOSVD_Reconstruction_99_Corrected.dat"
)

expected_bytes = int(np.prod(shape) * np.dtype(dtype).itemsize)

for path in [path_95, path_99]:
    if not os.path.exists(path):
        raise FileNotFoundError(path)
    if os.path.getsize(path) != expected_bytes:
        raise ValueError(f"Unexpected file size: {path}")

reconstruction_95 = np.memmap(
    path_95, dtype=dtype, mode="r", shape=shape
)
reconstruction_99 = np.memmap(
    path_99, dtype=dtype, mode="r", shape=shape
)

# --------------------------------------------------
# 2. LOAD TRACK METADATA
# --------------------------------------------------

track_metadata = pd.read_csv(
    os.path.join(BASE, "01_GT", "TRA", "man_track.txt"),
    sep=r"\s+",
    header=None,
    names=["Track_ID", "Start", "End", "Parent_ID"]
)

visible_tracks = visible_df["Track_ID"].astype(int).tolist()

# Cache masks so each required frame is read only once
mask_cache = {}

def get_track_mask(frame, track_id):
    if frame not in mask_cache:
        mask_path = os.path.join(
            BASE, "01_GT", "TRA",
            f"man_track{frame:03d}.tif"
        )

        if not os.path.exists(mask_path):
            mask_cache[frame] = None
        else:
            mask = tifffile.imread(mask_path)
            mask_cache[frame] = mask[:, 208:304, 306:402]

    cropped_mask = mask_cache[frame]

    if cropped_mask is None:
        return None

    return cropped_mask == track_id

# --------------------------------------------------
# 3. EXTRACT IDENTICAL FEATURES FROM EACH VERSION
# --------------------------------------------------

def extract_features_from_tensor(tensor, version):

    rows = []

    for track_id in visible_tracks:

        metadata = track_metadata[
            track_metadata["Track_ID"] == track_id
        ].iloc[0]

        start = int(metadata["Start"])
        end = int(metadata["End"])

        frame_means = []
        voxel_counts = []
        intensity_arrays = []

        for frame in range(start, end + 1):

            track_mask = get_track_mask(frame, track_id)

            if track_mask is None:
                continue

            voxel_count = int(np.count_nonzero(track_mask))

            if voxel_count == 0:
                continue

            image = np.asarray(tensor[frame], dtype=np.float32)
            intensities = image[track_mask]

            voxel_counts.append(voxel_count)
            frame_means.append(float(np.mean(intensities)))
            intensity_arrays.append(intensities)

        if not intensity_arrays:
            continue

        all_intensities = np.concatenate(intensity_arrays)
        frame_means = np.asarray(frame_means, dtype=np.float32)
        voxel_counts = np.asarray(voxel_counts, dtype=np.float32)

        if len(frame_means) > 1:
            temporal_change = float(
                np.mean(np.abs(np.diff(frame_means)))
            )
        else:
            temporal_change = 0.0

        label = int(
            svm_audit_df.loc[
                svm_audit_df["Track_ID"] == track_id,
                "Label"
            ].iloc[0]
        )

        rows.append({
            "Track_ID": track_id,
            "Label": label,
            "Duration": end - start + 1,
            "Observed_Frames": len(frame_means),
            "Mean_Intensity": float(np.mean(all_intensities)),
            "Std_Intensity": float(np.std(all_intensities)),
            "Min_Intensity": float(np.min(all_intensities)),
            "Max_Intensity": float(np.max(all_intensities)),
            "Mean_Frame_Intensity": float(np.mean(frame_means)),
            "Std_Frame_Intensity": float(np.std(frame_means)),
            "Mean_Temporal_Change": temporal_change,
            "Mean_Voxels_Per_Frame": float(np.mean(voxel_counts)),
            "Std_Voxels_Per_Frame": float(np.std(voxel_counts)),
            "Total_Observed_Voxels": int(np.sum(voxel_counts))
        })

    result = pd.DataFrame(rows)

    if result["Track_ID"].duplicated().any():
        raise ValueError("Duplicate track IDs found.")

    return result

# --------------------------------------------------
# 4. EXTRACT 95% AND 99% FEATURES
# --------------------------------------------------

features_95_df = extract_features_from_tensor(
    reconstruction_95, "HOSVD_95"
)

features_99_df = extract_features_from_tensor(
    reconstruction_99, "HOSVD_99"
)

# --------------------------------------------------
# 5. CHECK TRACK AND LABEL CONSISTENCY
# --------------------------------------------------

original_ids = set(visible_tracks)

for name, df in [
    ("95%", features_95_df),
    ("99%", features_99_df)
]:
    extracted_ids = set(df["Track_ID"].astype(int))

    if extracted_ids != original_ids:
        raise ValueError(
            f"{name} track IDs do not match the original dataset."
        )

    original_labels = (
        svm_audit_df.set_index("Track_ID")["Label"]
    )

    extracted_labels = df.set_index("Track_ID")["Label"]

    if not extracted_labels.equals(
        original_labels.loc[extracted_labels.index]
    ):
        raise ValueError(f"{name} labels do not match the audit.")

# --------------------------------------------------
# 6. SAVE FEATURE DATASETS
# --------------------------------------------------

save_95 = os.path.join(
    OUTPUT_DIR, "STEP35_SVM_Features_HOSVD_95.csv"
)
save_99 = os.path.join(
    OUTPUT_DIR, "STEP35_SVM_Features_HOSVD_99.csv"
)

features_95_df.to_csv(save_95, index=False)
features_99_df.to_csv(save_99, index=False)

# --------------------------------------------------
# 7. DISPLAY SUMMARY
# --------------------------------------------------

print("=" * 65)
print("STEP 35: HOSVD-SVM FEATURE EXTRACTION")
print("=" * 65)

print("Original tracks:", len(visible_tracks))
print("95% HOSVD tracks:", len(features_95_df))
print("99% HOSVD tracks:", len(features_99_df))

print("\n95% class distribution:")
print(features_95_df["Label"].value_counts().sort_index())

print("\n99% class distribution:")
print(features_99_df["Label"].value_counts().sort_index())

print("\n95% feature preview:")
display(features_95_df.head())

print("\n99% feature preview:")
display(features_99_df.head())

print("\nSaved files:")
print(save_95)
print(save_99)
```

```python

# STEP 35: PREPARE LEAKAGE-SAFE HOSVD-SVM BASELINE DATASET

import os
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# --------------------------------------------------
# 1. VERIFY INPUT DATA
# --------------------------------------------------

if "svm_features_df" not in globals():
    raise NameError("svm_features_df is missing. Run STEP 34 first.")

if "track_metadata" not in globals():
    raise NameError("track_metadata is missing. Run STEP 34 first.")

features_df = svm_features_df.copy()

feature_columns = [
    col for col in features_df.columns
    if col not in ["Track_ID", "Label"]
]

X = features_df[feature_columns].astype(float)
y = features_df["Label"].astype(int)

# --------------------------------------------------
# 2. BUILD LINEAGE GROUPS
# --------------------------------------------------

parent_lookup = dict(zip(
    track_metadata["Track_ID"].astype(int),
    track_metadata["Parent_ID"].astype(int)
))

def find_lineage_root(track_id):
    current = int(track_id)
    visited = set()

    while current in parent_lookup:
        if current in visited:
            raise ValueError(f"Cycle detected in lineage for track {track_id}")

        visited.add(current)
        parent = parent_lookup[current]

        if parent == 0:
            break

        current = parent

    return current

groups = features_df["Track_ID"].apply(find_lineage_root).to_numpy()

# --------------------------------------------------
# 3. FIND A GROUP-BASED TRAIN/TEST SPLIT
# --------------------------------------------------

# Try multiple deterministic group splits and select
# the one closest to the desired class proportions.

splitter = GroupShuffleSplit(
    n_splits=1000,
    test_size=0.25,
    random_state=42
)

overall_ratio = y.mean()
best_split = None
best_score = float("inf")

for train_idx, test_idx in splitter.split(X, y, groups):

    y_train_candidate = y.iloc[train_idx]
    y_test_candidate = y.iloc[test_idx]

    if y_train_candidate.nunique() < 2 or y_test_candidate.nunique() < 2:
        continue

    train_ratio = y_train_candidate.mean()
    test_ratio = y_test_candidate.mean()

    score = (
        abs(train_ratio - overall_ratio)
        + abs(test_ratio - overall_ratio)
        + abs(len(test_idx) / len(y) - 0.25)
    )

    if score < best_score:
        best_score = score
        best_split = (train_idx, test_idx)

if best_split is None:
    raise ValueError(
        "Could not find a group-based split containing both classes. "
        "Inspect lineage group distribution before proceeding."
    )

train_idx, test_idx = best_split

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

# --------------------------------------------------
# 4. VERIFY NO LINEAGE LEAKAGE
# --------------------------------------------------

train_groups = set(groups[train_idx])
test_groups = set(groups[test_idx])

overlap = train_groups.intersection(test_groups)

assert len(overlap) == 0, "Lineage leakage detected!"

# --------------------------------------------------
# 5. SCALE FEATURES USING TRAINING DATA ONLY
# --------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# --------------------------------------------------
# 6. FEATURE CORRELATION INSPECTION
# --------------------------------------------------

correlation_matrix = X.corr()

# --------------------------------------------------
# 7. SAVE PREPARED DATASET
# --------------------------------------------------

os.makedirs(OUTPUT_DIR, exist_ok=True)

split_df = features_df[["Track_ID", "Label"]].copy()
split_df["Lineage_Root"] = groups
split_df["Dataset_Split"] = "Unused"

split_df.loc[features_df.index[train_idx], "Dataset_Split"] = "Train"
split_df.loc[features_df.index[test_idx], "Dataset_Split"] = "Test"

split_df.to_csv(
    os.path.join(OUTPUT_DIR, "STEP35_SVM_Data_Split.csv"),
    index=False
)

np.savez(
    os.path.join(OUTPUT_DIR, "STEP35_SVM_Prepared_Data.npz"),
    X_train_scaled=X_train_scaled,
    X_test_scaled=X_test_scaled,
    y_train=y_train.to_numpy(),
    y_test=y_test.to_numpy(),
    feature_names=np.array(feature_columns)
)

# Save fitted scaler for later use
import joblib

joblib.dump(
    scaler,
    os.path.join(OUTPUT_DIR, "STEP35_SVM_Scaler.pkl")
)

# --------------------------------------------------
# 8. DISPLAY RESULTS
# --------------------------------------------------

print("=" * 65)
print("STEP 35: SVM DATASET PREPARATION")
print("=" * 65)

print("\nTotal samples:", len(features_df))
print("Numerical features:", len(feature_columns))
print("Unique lineage groups:", len(set(groups)))

print("\nTraining samples:", len(train_idx))
print("Testing samples:", len(test_idx))

print("\nTraining class distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting class distribution:")
print(y_test.value_counts().sort_index())

print("\nLineage groups in training:", len(train_groups))
print("Lineage groups in testing:", len(test_groups))
print("Overlapping lineage groups:", len(overlap))

print("\nFeature correlation matrix:")
display(correlation_matrix.round(2))

print("\nSplit summary:")
display(split_df)

print("\nSaved prepared dataset and scaler in:")
print(OUTPUT_DIR)
```

```python

# STEP 38B: TRAIN AND COMPARE ORIGINAL, HOSVD 95%, AND HOSVD 99% SVM MODELS

import os
import pandas as pd
import numpy as np
import joblib

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# --------------------------------------------------
# 1. LOAD THE THREE FEATURE DATASETS
# --------------------------------------------------

feature_files = {
    "Original": "STEP34_SVM_Track_Features.csv",
    "HOSVD_95": "STEP37_HOSVD_95_SVM_Features.csv",
    "HOSVD_99": "STEP37_HOSVD_99_SVM_Features.csv"
}

split_df = pd.read_csv(
    os.path.join(OUTPUT_DIR, "STEP35_SVM_Data_Split.csv")
)

results = []
trained_models = {}

# --------------------------------------------------
# 2. TRAIN AND EVALUATE EACH MODEL
# --------------------------------------------------

for model_name, filename in feature_files.items():

    print("\n" + "=" * 65)
    print(f"MODEL: {model_name}")
    print("=" * 65)

    feature_path = os.path.join(OUTPUT_DIR, filename)
    feature_df = pd.read_csv(feature_path)

    # Match features to the exact original split using Track_ID and Label
    merged = split_df[
        ["Track_ID", "Label", "Dataset_Split"]
    ].merge(
        feature_df,
        on=["Track_ID", "Label"],
        how="left",
        validate="one_to_one",
        indicator=True
    )

    # Verify that every split track has corresponding features
    assert len(merged) == len(split_df), "Track count mismatch!"
    assert (merged["_merge"] == "both").all(), "Missing feature rows!"

    merged = merged.drop(columns=["_merge"])

    feature_columns = [
        col for col in feature_df.columns
        if col not in ["Track_ID", "Label"]
    ]

    # Verify numerical features
    assert len(feature_columns) == 12
    assert merged[feature_columns].notnull().all().all()

    train_df = merged[merged["Dataset_Split"] == "Train"]
    test_df = merged[merged["Dataset_Split"] == "Test"]

    X_train = train_df[feature_columns]
    X_test = test_df[feature_columns]

    y_train = train_df["Label"]
    y_test = test_df["Label"]

    print("Training samples:", len(X_train))
    print("Testing samples:", len(X_test))
    print("Training class counts:")
    print(y_train.value_counts().sort_index().to_dict())
    print("Testing class counts:")
    print(y_test.value_counts().sort_index().to_dict())

    # Same SVM configuration for all three experiments
    model = Pipeline([
        ("scaler", StandardScaler()),
        ("svm", SVC(
            kernel="rbf",
            C=1,
            gamma="scale",
            class_weight="balanced"
        ))
    ])

    # Scaler is fitted only on training data through the pipeline
    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(
        y_test, predictions, zero_division=0
    )
    recall = recall_score(
        y_test, predictions, zero_division=0
    )
    f1 = f1_score(
        y_test, predictions, zero_division=0
    )

    cm = confusion_matrix(y_test, predictions, labels=[0, 1])

    print("\nEvaluation results:")
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-score:  {f1:.4f}")

    print("\nConfusion matrix:")
    print(cm)

    print("\nClassification report:")
    print(classification_report(
        y_test,
        predictions,
        labels=[0, 1],
        target_names=["No recorded children", "Division parent"],
        zero_division=0
    ))

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1,
        "TN": cm[0, 0],
        "FP": cm[0, 1],
        "FN": cm[1, 0],
        "TP": cm[1, 1]
    })

    trained_models[model_name] = model

    # Save each trained model
    model_path = os.path.join(
        OUTPUT_DIR,
        f"STEP38_{model_name}_SVM.pkl"
    )
    joblib.dump(model, model_path)

# --------------------------------------------------
# 3. DISPLAY FINAL COMPARISON
# --------------------------------------------------

comparison_df = pd.DataFrame(results)

print("\n" + "=" * 65)
print("SVM PERFORMANCE COMPARISON")
print("=" * 65)

print(comparison_df.to_string(index=False))

# Save comparison results
comparison_path = os.path.join(
    OUTPUT_DIR,
    "STEP38_SVM_Comparison.csv"
)

comparison_df.to_csv(comparison_path, index=False)

print("\nComparison saved to:")
print(comparison_path)

print("\nAll three SVM models saved successfully.")
```

```python
# STEP 39: ADAPTIVE HOSVD RANK SELECTION USING SPECTRAL ENTROPY

import os
import numpy as np
import pandas as pd

adaptive_results = []

for mode in range(1, 5):

    path = os.path.join(
        OUTPUT_DIR,
        f"singular_values_mode_{mode}.npy"
    )

    singular_values = np.load(path).astype(np.float64)

    # Convert singular values into normalized energy probabilities
    energy = singular_values ** 2
    probabilities = energy / np.sum(energy)

    # Normalized spectral entropy
    entropy = -np.sum(
        probabilities * np.log(probabilities + 1e-15)
    )

    normalized_entropy = entropy / np.log(len(singular_values))

    # Adaptive energy threshold
    threshold = 0.90 + 0.09 * normalized_entropy

    # Cumulative energy
    cumulative_energy = np.cumsum(probabilities)

    # Smallest rank satisfying the adaptive threshold
    adaptive_rank = np.searchsorted(
        cumulative_energy,
        threshold
    ) + 1

    adaptive_results.append({
        "Mode": mode,
        "Dimension": len(singular_values),
        "Normalized_Entropy": normalized_entropy,
        "Adaptive_Threshold": threshold,
        "Adaptive_Rank": adaptive_rank,
        "Achieved_Energy": cumulative_energy[adaptive_rank - 1]
    })

adaptive_rank_df = pd.DataFrame(adaptive_results)

print("=" * 65)
print("ADAPTIVE HOSVD RANK SELECTION")
print("=" * 65)

print(adaptive_rank_df.to_string(index=False))

adaptive_rank_path = os.path.join(
    OUTPUT_DIR,
    "STEP39_Adaptive_Rank_Selection.csv"
)

adaptive_rank_df.to_csv(adaptive_rank_path, index=False)

print("\nSaved:", adaptive_rank_path)

print("\nAdaptive rank tuple:")
print(tuple(adaptive_rank_df["Adaptive_Rank"]))
```

```python
# STEP 42: RECONSTRUCT ADAPTIVE HOSVD TENSOR

import os
import numpy as np

# Load adaptive components
core = np.load(
    os.path.join(OUTPUT_DIR, "HOSVD_Core_Adaptive.npy")
).astype(np.float32)

factors = np.load(
    os.path.join(OUTPUT_DIR, "HOSVD_Factors_Adaptive.npz")
)

U1 = factors["U1"].astype(np.float32)
U2 = factors["U2"].astype(np.float32)
U3 = factors["U3"].astype(np.float32)
U4 = factors["U4"].astype(np.float32)

print("=" * 65)
print("ADAPTIVE HOSVD RECONSTRUCTION")
print("=" * 65)

print("Core shape:", core.shape)

# Expand the width mode
temp = np.tensordot(
    core, U4, axes=([3], [1])
)
print("After width expansion:", temp.shape)

# Expand the height mode
temp = np.tensordot(
    U3, temp, axes=([1], [2])
)

# Reorder to (Rank T, Rank Z, Height, Width)
temp = np.transpose(temp, (1, 2, 0, 3))
print("After height expansion:", temp.shape)

# Expand the depth mode
temp = np.tensordot(
    U2, temp, axes=([1], [1])
)

# Reorder to (Rank T, Depth, Height, Width)
temp = np.transpose(temp, (1, 0, 2, 3))
print("After depth expansion:", temp.shape)

# Create disk-backed output
reconstruction_path = os.path.join(
    OUTPUT_DIR,
    "HOSVD_Reconstruction_Adaptive.dat"
)

adaptive_reconstruction = np.memmap(
    reconstruction_path,
    dtype=np.float32,
    mode="w+",
    shape=(250, 35, 96, 96)
)

# Expand time mode in blocks
block_size = 10

for start in range(0, 250, block_size):

    end = min(start + block_size, 250)

    block = np.tensordot(
        U1[start:end, :],
        temp,
        axes=([1], [0])
    )

    adaptive_reconstruction[start:end] = block

    print(f"Reconstructed frames {start} to {end - 1}")

    del block

adaptive_reconstruction.flush()

print("\nFinal reconstruction shape:",
      adaptive_reconstruction.shape)

print("Saved:", reconstruction_path)
```

```python
# STEP 43: COMPARE ADAPTIVE, 95%, AND 99% RECONSTRUCTION

import os
import numpy as np
import pandas as pd

# Load original pilot tensor
original = np.memmap(
    tensor_path,
    dtype=np.uint8,
    mode="r",
    shape=(250, 35, 96, 96)
)

reconstruction_paths = {
    "HOSVD_95": "HOSVD_Reconstruction_95_Corrected.dat",
    "HOSVD_99": "HOSVD_Reconstruction_99_Corrected.dat",
    "Adaptive": "HOSVD_Reconstruction_Adaptive.dat"
}

results = []

total_voxels = original.size
data_range = 243
block_size = 10

for model, filename in reconstruction_paths.items():

    print("\nEvaluating:", model)

    reconstruction = np.memmap(
        os.path.join(OUTPUT_DIR, filename),
        dtype=np.float32,
        mode="r",
        shape=(250, 35, 96, 96)
    )

    squared_error_sum = 0.0
    absolute_error_sum = 0.0
    original_squared_sum = 0.0

    # Global reconstruction metrics
    for start in range(0, 250, block_size):

        end = min(start + block_size, 250)

        x = np.asarray(original[start:end], dtype=np.float64)
        r = np.asarray(reconstruction[start:end], dtype=np.float64)

        error = r - x

        squared_error_sum += np.sum(error ** 2)
        absolute_error_sum += np.sum(np.abs(error))
        original_squared_sum += np.sum(x ** 2)

    mse = squared_error_sum / total_voxels
    rmse = np.sqrt(mse)
    mae = absolute_error_sum / total_voxels

    relative_frobenius = (
        np.sqrt(squared_error_sum) /
        np.sqrt(original_squared_sum)
    ) * 100

    psnr = 20 * np.log10(data_range / rmse)

    # Temporal metrics
    temporal_abs_change_sum = 0.0
    temporal_error_sum = 0.0
    temporal_voxels = (250 - 1) * 35 * 96 * 96

    for start in range(1, 250, block_size):

        end = min(start + block_size, 250)

        original_current = np.asarray(
            original[start:end], dtype=np.float64
        )
        original_previous = np.asarray(
            original[start-1:end-1], dtype=np.float64
        )

        recon_current = np.asarray(
            reconstruction[start:end], dtype=np.float64
        )
        recon_previous = np.asarray(
            reconstruction[start-1:end-1], dtype=np.float64
        )

        original_change = original_current - original_previous
        reconstructed_change = recon_current - recon_previous

        temporal_abs_change_sum += np.sum(
            np.abs(original_change)
        )

        temporal_error_sum += np.sum(
            np.abs(reconstructed_change - original_change)
        )

    mean_temporal_change = (
        temporal_abs_change_sum / temporal_voxels
    )

    temporal_mae = temporal_error_sum / temporal_voxels

    temporal_relative_error = (
        temporal_mae / mean_temporal_change
    ) * 100

    results.append({
        "Model": model,
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae,
        "Relative_Frobenius_Error_Percent": relative_frobenius,
        "PSNR_dB": psnr,
        "Mean_Temporal_Change": mean_temporal_change,
        "Temporal_MAE": temporal_mae,
        "Temporal_Relative_Error_Percent": temporal_relative_error
    })

    print("RMSE:", round(rmse, 4))
    print("MAE:", round(mae, 4))
    print("PSNR:", round(psnr, 4))
    print("Temporal MAE:", round(temporal_mae, 4))

    del reconstruction

# Display comparison
metrics_df = pd.DataFrame(results)

print("\n" + "=" * 85)
print("RECONSTRUCTION QUALITY COMPARISON")
print("=" * 85)

print(metrics_df.round(4).to_string(index=False))

# Save results
metrics_path = os.path.join(
    OUTPUT_DIR,
    "STEP43_Reconstruction_Comparison.csv"
)

metrics_df.to_csv(metrics_path, index=False)

print("\nComparison saved to:")
print(metrics_path)
```

```python

# STEP 48: SELECT ADAPTIVE HOSVD VERSION 2 RANKS

import os
import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. DEFINE ENERGY BOUNDS
# --------------------------------------------------

MIN_ENERGY = 0.95
MAX_ENERGY = 0.99

mode_names = {
    1: "Time",
    2: "Depth",
    3: "Height",
    4: "Width"
}

results = []

# --------------------------------------------------
# 2. CALCULATE ENTROPY AND ADAPTIVE RANKS
# --------------------------------------------------

for mode in range(1, 5):

    singular_path = os.path.join(
        OUTPUT_DIR,
        f"singular_values_mode_{mode}.npy"
    )

    if not os.path.exists(singular_path):
        raise FileNotFoundError(singular_path)

    singular_values = np.load(singular_path)

    energy = singular_values ** 2
    probabilities = energy / np.sum(energy)

    # Normalized spectral entropy
    nonzero = probabilities > 0

    entropy = -np.sum(
        probabilities[nonzero] *
        np.log(probabilities[nonzero])
    )

    normalized_entropy = entropy / np.log(len(probabilities))

    # Entropy-guided energy threshold
    target_energy = (
        MIN_ENERGY +
        (MAX_ENERGY - MIN_ENERGY) * normalized_entropy
    )

    cumulative_energy = np.cumsum(probabilities)

    adaptive_rank = int(
        np.searchsorted(
            cumulative_energy,
            target_energy
        ) + 1
    )

    achieved_energy = float(
        cumulative_energy[adaptive_rank - 1]
    )

    results.append({
        "Mode": mode,
        "Mode_Name": mode_names[mode],
        "Normalized_Entropy": normalized_entropy,
        "Target_Energy": target_energy,
        "Adaptive_Rank_V2": adaptive_rank,
        "Achieved_Energy": achieved_energy
    })

# --------------------------------------------------
# 3. CREATE RESULTS TABLE
# --------------------------------------------------

adaptive_v2_df = pd.DataFrame(results)

adaptive_v2_ranks = tuple(
    adaptive_v2_df["Adaptive_Rank_V2"].astype(int)
)

# --------------------------------------------------
# 4. VERIFY RANKS AGAINST EXISTING 99% RANKS
# --------------------------------------------------

rank_99 = (215, 28, 22, 90)

for i, (selected, maximum) in enumerate(
    zip(adaptive_v2_ranks, rank_99)
):
    if selected > maximum:
        raise ValueError(
            f"Mode {i+1}: Version 2 rank exceeds "
            "the available 99% factor rank."
        )

# --------------------------------------------------
# 5. SAVE RESULTS
# --------------------------------------------------

output_path = os.path.join(
    OUTPUT_DIR,
    "STEP48_Adaptive_Rank_Selection_V2.csv"
)

adaptive_v2_df.to_csv(output_path, index=False)

print("=" * 65)
print("STEP 48: ADAPTIVE HOSVD VERSION 2")
print("=" * 65)

print(adaptive_v2_df.to_string(index=False))

print("\nSelected Version 2 ranks:", adaptive_v2_ranks)

print("\nSaved:", output_path)

print("\nSTEP 48 COMPLETE")
```

```python

# STEP 52: COMPARE ALL HOSVD RECONSTRUCTION METHODS

import os
import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. DEFINE INPUT FILES
# --------------------------------------------------

T, Z, Y, X = 250, 35, 96, 96
shape = (T, Z, Y, X)

tensor_path = os.path.join(
    OUTPUT_DIR,
    "Sequence01_Pilot_4D_Tensor.dat"
)

reconstruction_files = {
    "HOSVD_95": "HOSVD_Reconstruction_95_Corrected.dat",
    "HOSVD_99": "HOSVD_Reconstruction_99_Corrected.dat",
    "Adaptive_V1": "HOSVD_Reconstruction_Adaptive.dat",
    "Adaptive_V2": "HOSVD_Reconstruction_Adaptive_V2.dat"
}

if not os.path.exists(tensor_path):
    raise FileNotFoundError(tensor_path)

original = np.memmap(
    tensor_path,
    dtype=np.uint8,
    mode="r",
    shape=shape
)

# --------------------------------------------------
# 2. DEFINE METRIC PARAMETERS
# --------------------------------------------------

DATA_RANGE = 243
BLOCK_SIZE = 16

results = []

# --------------------------------------------------
# 3. PROCESS EACH RECONSTRUCTION
# --------------------------------------------------

for model_name, filename in reconstruction_files.items():

    file_path = os.path.join(OUTPUT_DIR, filename)

    if not os.path.exists(file_path):
        raise FileNotFoundError(file_path)

    reconstruction = np.memmap(
        file_path,
        dtype=np.float32,
        mode="r",
        shape=shape
    )

    squared_error_sum = 0.0
    absolute_error_sum = 0.0
    original_squared_sum = 0.0
    voxel_count = 0

    temporal_error_sum = 0.0
    temporal_original_change_sum = 0.0
    temporal_count = 0

    previous_original = None
    previous_reconstruction = None

    # --------------------------------------------------
    # 4. CALCULATE GLOBAL AND TEMPORAL ERRORS
    # --------------------------------------------------

    for start in range(0, T, BLOCK_SIZE):

        end = min(start + BLOCK_SIZE, T)

        original_block = np.asarray(
            original[start:end],
            dtype=np.float64
        )

        reconstructed_block = np.asarray(
            reconstruction[start:end],
            dtype=np.float64
        )

        error = original_block - reconstructed_block

        squared_error_sum += np.sum(error ** 2)
        absolute_error_sum += np.sum(np.abs(error))
        original_squared_sum += np.sum(original_block ** 2)

        voxel_count += error.size

        # Temporal differences across block boundaries
        if previous_original is None:

            original_temporal = np.diff(
                original_block,
                axis=0
            )

            reconstructed_temporal = np.diff(
                reconstructed_block,
                axis=0
            )

        else:

            original_with_previous = np.concatenate(
                [previous_original[None, ...], original_block],
                axis=0
            )

            reconstruction_with_previous = np.concatenate(
                [
                    previous_reconstruction[None, ...],
                    reconstructed_block
                ],
                axis=0
            )

            original_temporal = np.diff(
                original_with_previous,
                axis=0
            )

            reconstructed_temporal = np.diff(
                reconstruction_with_previous,
                axis=0
            )

        temporal_difference_error = (
            original_temporal - reconstructed_temporal
        )

        temporal_error_sum += np.sum(
            np.abs(temporal_difference_error)
        )

        temporal_original_change_sum += np.sum(
            np.abs(original_temporal)
        )

        temporal_count += original_temporal.size

        previous_original = original_block[-1].copy()
        previous_reconstruction = reconstructed_block[-1].copy()

    # --------------------------------------------------
    # 5. CALCULATE FINAL METRICS
    # --------------------------------------------------

    mse = squared_error_sum / voxel_count
    rmse = np.sqrt(mse)
    mae = absolute_error_sum / voxel_count

    relative_frobenius = (
        np.sqrt(squared_error_sum) /
        np.sqrt(original_squared_sum)
    ) * 100

    psnr = 10 * np.log10(
        (DATA_RANGE ** 2) / mse
    )

    temporal_mae = (
        temporal_error_sum / temporal_count
    )

    mean_original_temporal_change = (
        temporal_original_change_sum / temporal_count
    )

    temporal_relative_error = (
        temporal_mae /
        mean_original_temporal_change
    ) * 100

    results.append({
        "Model": model_name,
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae,
        "Relative_Frobenius_Error_Percent": relative_frobenius,
        "PSNR_dB": psnr,
        "Temporal_MAE": temporal_mae,
        "Temporal_Relative_Error_Percent": temporal_relative_error
    })

    print(f"\nCompleted: {model_name}")
    print(f"RMSE: {rmse:.4f}")
    print(f"MAE: {mae:.4f}")
    print(f"PSNR: {psnr:.4f} dB")

    del reconstruction

# --------------------------------------------------
# 6. SAVE COMPARISON TABLE
# --------------------------------------------------

comparison_df = pd.DataFrame(results)

output_path = os.path.join(
    OUTPUT_DIR,
    "STEP52_All_Reconstruction_Comparison.csv"
)

comparison_df.to_csv(output_path, index=False)

# --------------------------------------------------
# 7. DISPLAY RESULTS
# --------------------------------------------------

print("\n" + "=" * 85)
print("STEP 52: COMPLETE RECONSTRUCTION COMPARISON")
print("=" * 85)

print(
    comparison_df.round(5).to_string(index=False)
)

print("\nSaved:", output_path)

print("\nSTEP 52 COMPLETE")
```

```python

# STEP 54: ADAPTIVE V2 DIVISION REGION EVALUATION

import numpy as np
import pandas as pd
import os

# Load original pilot tensor and Adaptive V2 reconstruction
original = np.memmap(
    os.path.join(OUTPUT_DIR, "Sequence01_Pilot_4D_Tensor.dat"),
    dtype=np.uint8, mode="r", shape=(250, 35, 96, 96)
)

adaptive_v2 = np.memmap(
    os.path.join(OUTPUT_DIR, "HOSVD_Reconstruction_Adaptive_V2.dat"),
    dtype=np.float32, mode="r", shape=(250, 35, 96, 96)
)

# Load the 14 verified division regions
regions = pd.read_csv(
    os.path.join(OUTPUT_DIR, "STEP30_Verified_Division_Regions.csv")
)

results = []

for _, region in regions.iterrows():

    event_frame = int(region["Event_Frame"])

    # Same three-frame event window used in previous evaluations
    start_frame = max(0, event_frame - 1)
    end_frame = min(249, event_frame + 1)

    z0 = int(region["Z_Min"])
    z1 = int(region["Z_Max"]) + 1
    y0 = int(region["Y0"])
    y1 = int(region["Y1"])
    x0 = int(region["X0"])
    x1 = int(region["X1"])

    original_roi = np.asarray(
        original[start_frame:end_frame + 1, z0:z1, y0:y1, x0:x1],
        dtype=np.float32
    )

    reconstructed_roi = np.asarray(
        adaptive_v2[start_frame:end_frame + 1, z0:z1, y0:y1, x0:x1],
        dtype=np.float32
    )

    error = original_roi - reconstructed_roi

    rmse = float(np.sqrt(np.mean(error ** 2)))
    mae = float(np.mean(np.abs(error)))

    results.append({
        "Parent_ID": int(region["Parent_ID"]),
        "Event_Frame": event_frame,
        "ROI_Voxels": int(original_roi.size),
        "Adaptive_V2_RMSE": rmse,
        "Adaptive_V2_MAE": mae
    })

# Save results
adaptive_v2_df = pd.DataFrame(results)

output_path = os.path.join(
    OUTPUT_DIR, "STEP54_Adaptive_V2_Division_Region_Metrics.csv"
)

adaptive_v2_df.to_csv(output_path, index=False)

print("=" * 85)
print("STEP 54: ADAPTIVE V2 DIVISION REGION RESULTS")
print("=" * 85)

print(adaptive_v2_df.to_string(index=False))

print("\nAverage per-event metrics:")
print("RMSE:", adaptive_v2_df["Adaptive_V2_RMSE"].mean())
print("MAE :", adaptive_v2_df["Adaptive_V2_MAE"].mean())

print("\nSaved:", output_path)
print("\nSTEP 54 COMPLETE")
```

```python

# STEP 56: EXTRACT SVM FEATURES FROM ADAPTIVE HOSVD V2 RECONSTRUCTION

import os
import numpy as np
import pandas as pd
import tifffile

# --------------------------------------------------
# 1. DEFINE INPUTS
# --------------------------------------------------

Y0, Y1 = 208, 304
X0, X1 = 306, 402

T, Z, Y, X = 250, 35, 96, 96

adaptive_v2_file = os.path.join(
    OUTPUT_DIR,
    "HOSVD_Reconstruction_Adaptive_V2.dat"
)

track_ids = svm_features_df["Track_ID"].astype(int).tolist()

label_lookup = dict(zip(
    svm_features_df["Track_ID"].astype(int),
    svm_features_df["Label"].astype(int)
))

# --------------------------------------------------
# 2. CHECK INPUT FILE
# --------------------------------------------------

if not os.path.exists(adaptive_v2_file):
    raise FileNotFoundError(
        f"Missing Adaptive V2 reconstruction: {adaptive_v2_file}"
    )

expected_size = T * Z * Y * X * np.dtype(np.float32).itemsize
actual_size = os.path.getsize(adaptive_v2_file)

if actual_size != expected_size:
    raise ValueError(
        f"Unexpected file size. Expected {expected_size}, "
        f"found {actual_size} bytes."
    )

# --------------------------------------------------
# 3. LOAD ADAPTIVE V2 RECONSTRUCTION
# --------------------------------------------------

adaptive_v2_reconstruction = np.memmap(
    adaptive_v2_file,
    dtype=np.float32,
    mode="r",
    shape=(T, Z, Y, X)
)

# --------------------------------------------------
# 4. EXTRACT FEATURES
# --------------------------------------------------

rows = []

for track_id in track_ids:

    metadata = track_metadata[
        track_metadata["Track_ID"] == track_id
    ].iloc[0]

    start = int(metadata["Start"])
    end = int(metadata["End"])

    frame_means = []
    frame_voxel_counts = []
    intensity_arrays = []

    for frame in range(start, end + 1):

        mask_path = os.path.join(
            BASE,
            "01_GT",
            "TRA",
            f"man_track{frame:03d}.tif"
        )

        if not os.path.exists(mask_path):
            continue

        mask = tifffile.imread(mask_path)
        cropped_mask = mask[:, Y0:Y1, X0:X1]

        track_mask = cropped_mask == track_id
        voxel_count = np.count_nonzero(track_mask)

        if voxel_count == 0:
            continue

        image = np.asarray(
            adaptive_v2_reconstruction[frame],
            dtype=np.float32
        )

        intensities = image[track_mask]

        frame_voxel_counts.append(int(voxel_count))
        frame_means.append(float(np.mean(intensities)))
        intensity_arrays.append(intensities)

    if not intensity_arrays:
        continue

    all_intensities = np.concatenate(intensity_arrays)

    frame_means = np.asarray(frame_means, dtype=np.float32)
    voxel_counts = np.asarray(frame_voxel_counts, dtype=np.float32)

    if len(frame_means) > 1:
        temporal_change = float(
            np.mean(np.abs(np.diff(frame_means)))
        )
    else:
        temporal_change = 0.0

    rows.append({
        "Track_ID": track_id,
        "Label": label_lookup[track_id],
        "Duration": int(end - start + 1),
        "Observed_Frames": len(frame_means),
        "Mean_Intensity": float(np.mean(all_intensities)),
        "Std_Intensity": float(np.std(all_intensities)),
        "Min_Intensity": float(np.min(all_intensities)),
        "Max_Intensity": float(np.max(all_intensities)),
        "Mean_Frame_Intensity": float(np.mean(frame_means)),
        "Std_Frame_Intensity": float(np.std(frame_means)),
        "Mean_Temporal_Change": temporal_change,
        "Mean_Voxels_Per_Frame": float(np.mean(voxel_counts)),
        "Std_Voxels_Per_Frame": float(np.std(voxel_counts)),
        "Total_Observed_Voxels": int(np.sum(voxel_counts))
    })

adaptive_v2_features_df = pd.DataFrame(rows)

# --------------------------------------------------
# 5. SAVE FEATURES
# --------------------------------------------------

adaptive_v2_output_path = os.path.join(
    OUTPUT_DIR,
    "STEP56_SVM_Adaptive_V2_Features.csv"
)

adaptive_v2_features_df.to_csv(
    adaptive_v2_output_path,
    index=False
)

# --------------------------------------------------
# 6. VERIFY TRACKS AND FEATURES
# --------------------------------------------------

expected_ids = set(track_ids)

actual_ids = set(
    adaptive_v2_features_df["Track_ID"].astype(int)
)

if actual_ids != expected_ids:
    raise ValueError(
        f"Track mismatch. Missing: {expected_ids - actual_ids}, "
        f"Unexpected: {actual_ids - expected_ids}"
    )

if adaptive_v2_features_df.isnull().any().any():
    raise ValueError("Missing feature values detected.")

if len(adaptive_v2_features_df.columns) != 14:
    raise ValueError("Unexpected number of columns.")

print("=" * 65)
print("STEP 56 COMPLETE")
print("=" * 65)

print("Tracks extracted:", len(adaptive_v2_features_df))
print("Missing tracks:", len(expected_ids - actual_ids))
print("Feature columns:", len(adaptive_v2_features_df.columns) - 2)

print("\nClass distribution:")
print(adaptive_v2_features_df["Label"].value_counts().sort_index())

print("\nFeature preview:")
print(adaptive_v2_features_df.head().to_string(index=False))

print("\nSaved:", adaptive_v2_output_path)

del adaptive_v2_reconstruction
```

```python

# STEP 58: FINAL COMPARISON OF ALL FIVE SVM MODELS

import os
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. LOAD SAVED SVM RESULTS
# --------------------------------------------------

files = {
    "Original": "STEP36_SVM_Metrics.csv",
    "HOSVD_95": "STEP38_SVM_Comparison.csv",
    "Adaptive_V1": "STEP47_Adaptive_SVM_Metrics.csv",
    "Adaptive_V2": "STEP57_Adaptive_V2_SVM_Metrics.csv"
}

loaded = {}

for name, filename in files.items():
    path = os.path.join(OUTPUT_DIR, filename)

    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing result file: {path}")

    loaded[name] = pd.read_csv(path)

# --------------------------------------------------
# 2. NORMALIZE METRIC COLUMN NAMES
# --------------------------------------------------

def normalize_metrics(df, model_name=None):

    df = df.copy()

    # Find metric columns without assuming exact capitalization
    normalized = {
        str(col).strip().lower().replace(" ", "_"): col
        for col in df.columns
    }

    required = {
        "accuracy": ["accuracy"],
        "precision": ["precision"],
        "recall": ["recall"],
        "f1": ["f1", "f1_score", "f1-score"]
    }

    selected = {}

    for metric, possible_names in required.items():
        found = None

        for candidate in possible_names:
            if candidate in normalized:
                found = normalized[candidate]
                break

        if found is None:
            raise ValueError(
                f"Cannot find {metric} column. "
                f"Available columns: {list(df.columns)}"
            )

        selected[metric] = found

    # Select a specific model row if the file contains multiple models
    if model_name is not None and "model" in normalized:
        model_col = normalized["model"]

        matching = df[
            df[model_col].astype(str).str.contains(
                model_name, case=False, regex=False
            )
        ]

        if len(matching) == 0:
            raise ValueError(
                f"Model {model_name} not found in results."
            )

        df = matching.iloc[[0]]

    else:
        df = df.iloc[[0]]

    return {
        "Accuracy": float(df.iloc[0][selected["accuracy"]]),
        "Precision": float(df.iloc[0][selected["precision"]]),
        "Recall": float(df.iloc[0][selected["recall"]]),
        "F1_Score": float(df.iloc[0][selected["f1"]])
    }

# --------------------------------------------------
# 3. EXTRACT THE THREE MODELS FROM STEP 38
# --------------------------------------------------

step38 = loaded["HOSVD_95"]

hosvd95 = normalize_metrics(step38, "HOSVD_95")
hosvd99 = normalize_metrics(step38, "HOSVD_99")

# Original baseline comes from STEP 36
original = normalize_metrics(loaded["Original"])

# Adaptive models
adaptive_v1 = normalize_metrics(loaded["Adaptive_V1"])
adaptive_v2 = normalize_metrics(loaded["Adaptive_V2"])

# --------------------------------------------------
# 4. BUILD FINAL COMPARISON TABLE
# --------------------------------------------------

final_results = pd.DataFrame([
    {"Model": "Original + SVM", **original},
    {"Model": "HOSVD 95% + SVM", **hosvd95},
    {"Model": "HOSVD 99% + SVM", **hosvd99},
    {"Model": "Adaptive V1 + SVM", **adaptive_v1},
    {"Model": "Adaptive V2 + SVM", **adaptive_v2}
])

# Convert metrics to percentages for display
percentage_results = final_results.copy()

metric_columns = [
    "Accuracy", "Precision", "Recall", "F1_Score"
]

percentage_results[metric_columns] *= 100

# --------------------------------------------------
# 5. DISPLAY RESULTS
# --------------------------------------------------

print("=" * 90)
print("STEP 58: FINAL FIVE-MODEL SVM COMPARISON")
print("=" * 90)

print(percentage_results.round(2).to_string(index=False))

# Identify best model for each metric
print("\nBEST MODEL FOR EACH METRIC")

for metric in metric_columns:
    best_index = final_results[metric].idxmax()

    print(
        f"{metric}: "
        f"{final_results.loc[best_index, 'Model']} "
        f"({final_results.loc[best_index, metric] * 100:.2f}%)"
    )

# --------------------------------------------------
# 6. SAVE FINAL TABLE
# --------------------------------------------------

csv_path = os.path.join(
    OUTPUT_DIR, "STEP58_Final_Five_Model_SVM_Comparison.csv"
)

percentage_results.to_csv(csv_path, index=False)

# --------------------------------------------------
# 7. GENERATE COMPARISON GRAPH
# --------------------------------------------------

plot_df = percentage_results.set_index("Model")

ax = plot_df.plot(
    kind="bar",
    figsize=(12, 6)
)

ax.set_title("Comparison of Five SVM Models")
ax.set_xlabel("Model")
ax.set_ylabel("Performance (%)")
ax.set_ylim(0, 100)
ax.legend(title="Metrics")
ax.tick_params(axis="x", rotation=20)

plt.tight_layout()

plot_path = os.path.join(
    OUTPUT_DIR, "STEP58_Five_Model_SVM_Comparison.png"
)

plt.savefig(plot_path, dpi=300, bbox_inches="tight")
plt.show()

print("\nSaved comparison table:", csv_path)
print("Saved comparison graph:", plot_path)

print("\nSTEP 58 COMPLETE")
```
